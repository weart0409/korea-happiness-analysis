## 2. 시계열 점수 및 순위 추이 시각화

```python
fig, ax1 = plt.subplots(figsize=(10, 5))

ax1.plot(korea_df['Year'], korea_df['Life evaluation (3-year average)'], marker='o', color='#2563eb', linewidth=2.5, label='행복 점수')
ax1.set_xlabel('연도 (Year)', fontsize=12)
ax1.set_ylabel('Cantril Ladder (0-10)', color='#2563eb', fontsize=12)
ax1.set_ylim(5.0, 7.0)
ax1.grid(True, linestyle='--', alpha=0.5)

ax2 = ax1.twinx()
ax2.plot(korea_df['Year'], korea_df['Rank'], marker='s', color='#dc2626', linewidth=2, linestyle='--', label='글로벌 순위')
ax2.set_ylabel('글로벌 순위 (Rank)', color='#dc2626', fontsize=12)
ax2.invert_yaxis()
ax2.set_ylim(80, 30)

plt.title('대한민국 행복 점수 및 순위 추이 (2011-2025)', fontsize=14, fontweight='bold')
plt.show()
```
